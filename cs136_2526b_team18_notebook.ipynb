We first start by describing the forward kinematics of the two-link robot arm.

The base joint is fixed at the origin $O_1$. Each link is placed end-to-end, so we apply a chain of 2D rotations.

The first link has length $L_1$ and makes angle $\theta_1$ with the positive $x$-axis. By trigonometry, the elbow (end of link 1) is at

$$
x_1 = L_1 \cos \theta_1,
\qquad
y_1 = L_1 \sin \theta_1
$$

The second link has length $L_2$. Its angle $\theta_2$ is measured relative to the first link, so its absolute angle with respect to the $x$-axis is $(\theta_1 + \theta_2)$.

Adding the displacement of link 2 starting from the elbow, we have

$$
x_E = x_1 + L_2 \cos(\theta_1 + \theta_2)
$$

$$
y_E = y_1 + L_2 \sin(\theta_1 + \theta_2)
$$

Substituting the expressions for $x_1$ and $y_1$ gives

$$
x_E = L_1 \cos \theta_1 + L_2 \cos(\theta_1 + \theta_2)
$$

$$
y_E = L_1 \sin \theta_1 + L_2 \sin(\theta_1 + \theta_2)
$$


**Final answer:**
$$
\boxed{
\left\langle
L_1 \cos \theta_1 + L_2 \cos(\theta_1 + \theta_2),
\;
L_1 \sin \theta_1 + L_2 \sin(\theta_1 + \theta_2)
\right\rangle
}
$$

---

Using SymPy, write the expression here:

````python
L1, L2, t = sp.symbols('L1 L2 t', positive=True, real=True)
theta1 = sp.Function('theta_1')(t)
theta2 = sp.Function('theta_2')(t)

# Give the forward kinematics expression here:
x_sym = L1*sp.cos(theta1) + L2*sp.cos(theta1+theta2)
y_sym = L1*sp.sin(theta1) + L2*sp.sin(theta1+theta2)

display(Math('x(t) = ' + sp.latex(x_sym)))
display(Math('y(t) = ' + sp.latex(y_sym)))
````

Since you will be using this frequently, we can use a helper function `compute_tip(l1, l2, t1, t2)` that returns the coordinates as a NumPy array `[x_E, y_E]`. When `t1` and `t2` are arrays, transposing the result gives an array of shape `(n_tips,2)`.

We also need to reuse the expressions `x_sym` and `y_sym`, thus we can turn that expression into lambda functions `x_func` and `y_func`, respectively. This means that if we insert NumPy arrays onto `t1` and `t2`, it should output a respective NumPy array where its values were computed using the formula defined in their respective SymPy expressions.
